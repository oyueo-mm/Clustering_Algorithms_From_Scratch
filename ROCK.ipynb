```python
import numpy as np

# 데이터는 학과 | 학번 | 성별로 만들어봄
data = [
    {"인공지능", "26", "남자"},
    {"데이터사이언스", "26", "여자"},
    {"데이터사이언스", "26", "남자"},
    {"응용소프트웨어", "25", "남자"},
    {"응용소프트웨어", "25", "여자"},
    {"인공지능", "24", "남자"},
    {"데이터사이언스", "24", "남자"},
    {"데이터사이언스", "24", "여자"},
    {"응용소프트웨어", "23", "남자"},
    {"인공지능", "22", "남자"},
    {"인공지능", "22", "여자"},
    {"응용소프트웨어", "22", "여자"}
]

def sim(a, b):
  return len(a.intersection(b)) / len(a.union(b))

theta = 0.3
adjacency_matrix = np.zeros((len(data), len(data)))
# adjacency_matrix는 인접 행렬임. 두 데이터의 관계를 나타냄. 1은 이웃, 0은 이웃 아님

for i in range(len(data)):
  for j in range(len(data)):
    if sim(data[i], data[j]) > theta:
      adjacency_matrix[i][j] = 1

link_matrix = np.dot(adjacency_matrix, adjacency_matrix)


def goodness_measure(C_i, C_j):
  cross_link = 0
  for i in C_i:
    for j in C_j:
      cross_link += link_matrix[i][j]

  f_theta = (1 + theta) / (1 - theta)
  exponent = 1 + 2 * f_theta
  return cross_link / (((len(C_i) + len(C_j))**exponent) - (len(C_i))**exponent - (len(C_j))**exponent)
```

```python
clusters = [[x] for x in range(len(data))]
k = 3

while len(clusters) > k:
  best_score = -1
  best_pair = (None, None)
  for i in range(len(clusters)):
    for j in range(i + 1, len(clusters)):
      score = goodness_measure(clusters[i], clusters[j])
      if score > best_score:
        best_score = score
        best_pair = (i, j)
  # 더 이상 합칠 게 없으면 폭파
  if best_score <= 0:
    break
  new_cluster = clusters[best_pair[0]] + clusters[best_pair[1]]
  clusters.pop(best_pair[0])
  clusters.pop(best_pair[1])
  clusters.append(new_cluster)
```

```python
print("=======================================")
for i, D in enumerate(clusters):
  print(f"클러스터{i+1} :{D}")
  for j in D:
    print(data[j])
  print("=======================================")
```
